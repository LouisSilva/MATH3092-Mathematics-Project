```python
%load_ext autoreload
%autoreload 2

import shutil
import os
from sklearn.model_selection import train_test_split

from src.fma_utils import create_dataset_df

from src.TestSuite.data_structures import BenchmarkConfig
from src.TestSuite.benchmark_runner import BenchmarkRunner
from src.TestSuite.test_cases import CleanTrackTest, WhiteNoiseTest, PitchShiftTest, ReverbTest, ClippingDbTest, DistortionTest

from src.PCAFingerprinter.pca_fingerprinter import PCAFingerprinter, DeltaFingerprint
from src.PCAFingerprinter.binary_fingerprint_search import HammingSearch
from src.retrieval_backends import PCARetrievalBackend, ShazamRetrievalBackend
```

```python
METADATA_DIR = "B:\\Documents\\Uni\\MATH3092 Mathematics Project\\MusicDatasets\\fma\\fma_metadata"
AUDIO_FILES_DIR = "B:\\Documents\\Uni\\MATH3092 Mathematics Project\\MusicDatasets\\fma\\fma_small\\fma_small"

RUNNER_STATE_PATH = "./data/runner_state.json"
PCA_DB_PATH = "./data/pca_db.npz"
SHAZAM_DB_PATH = "./data/shazam_db.pklz"
PCA_MODEL_PATH = "./data/pca_model.joblib"

tracks, my_songs = create_dataset_df(METADATA_DIR, AUDIO_FILES_DIR)
```

```python
random_seed = 42

X_train, X_test = train_test_split(
    tracks,
    test_size=0.9,
    random_state=random_seed
)

# --------------------------------------------------------
pca_fingerprinter = PCAFingerprinter(
    fingerprint_strategy=DeltaFingerprint(),
    f_s=44100,
    L=4096,
    H=2048, # L // 2
    B=128,
    tau=80,
    window_type="hann",
    gamma=500000,
    kappa=500,
    phi=16,
    random_state=random_seed
)

pca_backend = PCARetrievalBackend(
    fingerprinter=pca_fingerprinter,
    search_strategy=HammingSearch(),
    confident_match_threshold=0.4,
)
# --------------------------------------------------------
shazam_backend = ShazamRetrievalBackend(
    density=20.0,
    samplerate=11025,
    n_fft=512,
    shifts=4,
    pks_per_frame=5,
    fanout=3,
    freq_sd=30.0,
    match_win=1,
    min_count=0,
    max_matches=1,
    search_depth=100,
    sort_by_time=False,
    exact_count=False,
    hashbits=20,
    depth=100,
    maxtime=16384,
    verbose=False,
    confident_match_threshold=12,
)
# --------------------------------------------------------

# Create test config
benchmark_config = BenchmarkConfig(
    f_s=44100,
    snippet_duration_sec=5.0,
    num_db_tracks=6000,
    num_positive_queries=2000,
    num_negative_queries=200,
    random_seed=random_seed,
    save_failed_queries=False,
    save_passed_queries=False,
    temp_dir="temp_queries/multi_backend"
)

benchmark_runner = BenchmarkRunner(
    backends=[pca_backend, shazam_backend],
    all_filepaths=X_test['filepath'].tolist(),
    config=benchmark_config
)
```

```python
# Check if we have saved data
if os.path.exists(RUNNER_STATE_PATH) and os.path.exists(PCA_DB_PATH) and os.path.exists(SHAZAM_DB_PATH):
    print("Loading saved state and databases...")

    # Load PCA model
    pca_fingerprinter = PCAFingerprinter.load_model(PCA_MODEL_PATH)
    pca_backend.fingerprinter = pca_fingerprinter # Ensure backend uses the loaded model

    # Load runner state (skips setup_database)
    #benchmark_runner.load_state(RUNNER_STATE_PATH)
    benchmark_runner.load_restricted_state(RUNNER_STATE_PATH)

    # Load backend databases
    pca_backend.load_database(PCA_DB_PATH)
    shazam_backend.load_database(SHAZAM_DB_PATH)

else:
    print("No saved state found. Generating from scratch...")

    # Train and save PCA
    pca_fingerprinter.train(X_train['filepath'].tolist())
    pca_fingerprinter.save_model(PCA_MODEL_PATH)

    # Setup database (scans files, selects tracks, populates backends)
    benchmark_runner.setup_database()

    # Save everything for next time
    benchmark_runner.save_state(RUNNER_STATE_PATH)
    pca_backend.save_database(PCA_DB_PATH)
    shazam_backend.save_database(SHAZAM_DB_PATH)
```

```python
test_cases_to_run =[
    CleanTrackTest(),
    WhiteNoiseTest(snr_db=10),
    WhiteNoiseTest(snr_db=0),
    WhiteNoiseTest(snr_db=-10),
    PitchShiftTest(n_steps=1),
    PitchShiftTest(n_steps=2),
    DistortionTest(drive_db=20),
    ReverbTest(room_size=0.85, damping=0.6, wet_level=0.7, dry_level=0.3),
    ClippingDbTest(threshold_db=-20),
]

# Run the experiment
# benchmark_runner.run_experiment(test_cases_to_run)
benchmark_runner.run_experiment_concurrently(test_cases=test_cases_to_run, max_workers=8)

results_df = benchmark_runner.get_results_df()
benchmark_runner.summarize_results()
```
