```python
%load_ext autoreload
%autoreload 2

import shutil
from sklearn.model_selection import train_test_split

from src.fma_utils import create_dataset_df

from src.TestSuite.data_structures import BenchmarkConfig
from src.TestSuite.benchmark_runner import BenchmarkRunner
from src.TestSuite.test_cases import CleanTrackTest, WhiteNoiseTest, PitchShiftTest, ReverbTest, ClippingDbTest, DistortionTest

from src.EigenSpectraFingerprinter.pca_fingerprinter import PCAFingerprinter, DeltaFingerprint
from src.EigenSpectraFingerprinter.binary_fingerprint_search import HammingSearch, FFTConvolveSearch
from src.retrieval_backends import PCARetrievalBackend, ShazamRetrievalBackend
```

```python
METADATA_DIR = "B:\\Documents\\Uni\\MATH3092 Mathematics Project\\MusicDatasets\\fma\\fma_metadata"
USER_SONGS_DIR = "B:\\Documents\\Uni\\MATH3092 Mathematics Project\\MusicDatasets\\my_songs" # random extra songs that aren't part of the FMA dataset
AUDIO_FILES_DIR = "B:\\Documents\\Uni\\MATH3092 Mathematics Project\\MusicDatasets\\fma\\fma_small\\fma_small"

tracks, my_songs = create_dataset_df(METADATA_DIR, AUDIO_FILES_DIR, USER_SONGS_DIR)
```

```python
f_s = 44100
L = 2048 * 2
H = L // 2
window_type = "hann"
B = 128
tau = 70
gamma = 50000
kappa = 500
phi = 16
random_seed = 42
match_threshold = 0.35

test_size = 0.5
num_tracks = 8000

reduced_track_db = tracks.iloc[0:num_tracks]
X_train, X_test = train_test_split(
    reduced_track_db,
    test_size=test_size,
    random_state=random_seed
)

# Create the fingerprinter object
pca_fingerprinter = PCAFingerprinter(
    fingerprint_strategy=DeltaFingerprint(),
    f_s=f_s,
    L=L,
    H=H,
    B=B,
    tau=tau,
    window_type=window_type,
    gamma=gamma,
    kappa=kappa,
    phi=phi,
    random_state=random_seed
)

pca_backend = PCARetrievalBackend(
    fingerprinter=pca_fingerprinter,
    search_strategy=FFTConvolveSearch(),
    confident_match_threshold=0.35,
)

# Create the test config
pca_config = BenchmarkConfig(
    f_s=f_s,
    snippet_duration_sec=5.0,
    n_db_tracks=100,
    n_query_tracks=50,
    random_seed=random_seed,
    save_failed_queries=True,
    save_passed_queries=True,
    temp_dir="temp_queries/pca"
)

# Create the experiment runner
pca_runner = BenchmarkRunner(
    backend=pca_backend,
    all_filepaths=X_test['filepath'].tolist(),
    config=pca_config
)

# ------- Shazam -------

shazam_backend = ShazamRetrievalBackend(
    density=20.0,
    samplerate=11025,
    n_fft=512,
    shifts=4,
    pks_per_frame=5,
    fanout=3,
    freq_sd=30.0,
    match_win=1,
    min_count=5,
    max_matches=1,
    search_depth=100,
    sort_by_time=False,
    exact_count=False,
    hashbits=20,
    depth=100,
    maxtime=16384,
    verbose=False,
    confident_match_threshold=6,
)

# Create the test config
shazam_config = BenchmarkConfig(
    f_s=f_s,
    snippet_duration_sec=5.0,
    n_db_tracks=100,
    n_query_tracks=50,
    random_seed=random_seed,
    save_failed_queries=True,
    save_passed_queries=True,
    temp_dir="temp_queries/shazam"
)

# Create the experiment runner
shazam_runner = BenchmarkRunner(
    backend=shazam_backend,
    all_filepaths=X_test['filepath'].tolist(),
    config=shazam_config
)
```

```python
pca_runner.train(X_train['filepath'].tolist())
```

```python
# pca_runner.setup_database()
shazam_runner.setup_database()
```

```python
test_cases_to_run = [
    # CleanTrackTest(),
    # WhiteNoiseTest(amplitude=0.05),
    # WhiteNoiseTest(amplitude=0.10),
    WhiteNoiseTest(amplitude=0.15),
    # PitchShiftTest(n_steps=0.5), # Shift up by half a semitone
    DistortionTest(drive_db=20),
    ReverbTest(room_size=0.85, damping=0.6, wet_level=0.7, dry_level=0.3),
    ClippingDbTest(threshold_db=-12),
]

pca_runner.run_experiment(test_cases=test_cases_to_run)
results_df = pca_runner.get_results_df()

print("\n--- DETAILED RESULTS ---")
print(results_df.to_string())

pca_runner.summarize_results() # 049029
```
