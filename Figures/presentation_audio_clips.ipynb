```python
import os
import numpy as np
from pedalboard import PitchShift, Clipping, Reverb
import soundfile as sf

from src.fma_utils import create_dataset_df
from src.PCAFingerprinter.spectrogram_generation import load_audio
```

```python
METADATA_DIR = "B:\\Documents\\Uni\\MATH3092 Mathematics Project\\MusicDatasets\\fma\\fma_metadata"
AUDIO_FILES_DIR = "B:\\Documents\\Uni\\MATH3092 Mathematics Project\\MusicDatasets\\fma\\fma_small\\fma_small"
USER_SONGS_DIR = "B:\\Documents\\Uni\\MATH3092 Mathematics Project\\MusicDatasets\\my_songs"  # random extra songs that aren't part of the FMA dataset

tracks, my_songs = create_dataset_df(METADATA_DIR, AUDIO_FILES_DIR, USER_SONGS_DIR)
```

```python
song_filepath = my_songs.loc[my_songs["track_title"] == "14 - Ball-istics Outro.flac", "filepath"].iloc[0]

f_s = 44100
SNR = 0
n_steps = 2
threshold_db = -20
room_size = 0.85
damping = 0.6
wet_level = 0.7
dry_level = 0.3

x, _ = load_audio(song_filepath, f_s=f_s, offset=10, duration=6)

# --- Gaussian white noise ---
average_x_power = np.mean(x ** 2) + 1e-12
average_omega_power = average_x_power / (10 ** (SNR / 10))

omega_standard_deviation = np.sqrt(average_omega_power)
omega = np.random.normal(loc=0, scale=omega_standard_deviation, size=x.shape[0]).astype(x.dtype)

y_white_noise = x + omega

# --- Pitch shift ---
y_pitch_shift = PitchShift(semitones=n_steps)(x, f_s)

# --- Clipping ---
y_clipped = Clipping(threshold_db=threshold_db)(x, f_s)

# --- Reverb ---
y_reverb = Reverb(
    room_size=room_size,
    damping=damping,
    wet_level=wet_level,
    dry_level=dry_level
)(x, f_s)
```

```python
os.makedirs("./temp", exist_ok=True)

sf.write("./temp/ballistics_clean.wav", x, f_s)
sf.write("./temp/ballistics_white_noise.wav", y_white_noise, f_s)
sf.write("./temp/ballistics_pitch_shift.wav", y_pitch_shift, f_s)
sf.write("./temp/ballistics_clipped.wav", y_clipped, f_s)
sf.write("./temp/ballistics_reverb.wav", y_reverb, f_s)
```
