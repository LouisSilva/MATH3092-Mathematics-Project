```python
%load_ext autoreload
%autoreload 2

import shutil
from sklearn.model_selection import train_test_split

from src.fma_utils import create_dataset_df

from src.TestSuite.data_structures import BenchmarkConfig
from src.TestSuite.benchmark_runner import BenchmarkRunner
from src.TestSuite.test_cases import CleanTrackTest, WhiteNoiseTest, PitchShiftTest, ReverbTest, ClippingDbTest, DistortionTest

from src.EigenSpectraFingerprinter.pca_fingerprinter import PCAFingerprinter, DeltaFingerprint
from src.EigenSpectraFingerprinter.binary_fingerprint_search import HammingSearch, FFTConvolveSearch
from src.retrieval_backends import PCARetrievalBackend, ShazamRetrievalBackend
```

```python
METADATA_DIR = "B:\\Documents\\Uni\\MATH3092 Mathematics Project\\MusicDatasets\\fma\\fma_metadata"
USER_SONGS_DIR = "B:\\Documents\\Uni\\MATH3092 Mathematics Project\\MusicDatasets\\my_songs" # random extra songs that aren't part of the FMA dataset
AUDIO_FILES_DIR = "B:\\Documents\\Uni\\MATH3092 Mathematics Project\\MusicDatasets\\fma\\fma_small\\fma_small"

tracks, my_songs = create_dataset_df(METADATA_DIR, AUDIO_FILES_DIR, USER_SONGS_DIR)
```

```python
random_seed = 42

X_train, X_test = train_test_split(
    tracks,
    test_size=0.9,
    random_state=random_seed
)

# --------------------------------------------------------
pca_fingerprinter = PCAFingerprinter(
    fingerprint_strategy=DeltaFingerprint(),
    f_s=44100,
    L=4096,
    H=2048, # L // 2
    B=128,
    tau=80,
    window_type="hann",
    gamma=50000,
    kappa=500,
    phi=16,
    random_state=random_seed
)

pca_backend = PCARetrievalBackend(
    fingerprinter=pca_fingerprinter,
    search_strategy=FFTConvolveSearch(),
    confident_match_threshold=0.35,
)
# --------------------------------------------------------
shazam_backend = ShazamRetrievalBackend(
    density=20.0,
    samplerate=11025,
    n_fft=512,
    shifts=4,
    pks_per_frame=5,
    fanout=3,
    freq_sd=30.0,
    match_win=1,
    min_count=5,
    max_matches=1,
    search_depth=100,
    sort_by_time=False,
    exact_count=False,
    hashbits=20,
    depth=100,
    maxtime=16384,
    verbose=False,
    confident_match_threshold=6,
)
# --------------------------------------------------------

# Create test config
benchmark_config = BenchmarkConfig(
    f_s=44100,
    snippet_duration_sec=5.0,
    n_db_tracks=100,
    n_query_tracks=50,
    random_seed=random_seed,
    save_failed_queries=True,
    save_passed_queries=True,
    temp_dir="temp_queries/multi_backend"
)

benchmark_runner = BenchmarkRunner(
    backends=[pca_backend, shazam_backend],
    all_filepaths=X_test['filepath'].tolist(),
    config=benchmark_config
)
```

```python
benchmark_runner.train(X_train['filepath'].tolist())
```

```python
benchmark_runner.setup_database()
```

```python
test_cases_to_run =[
    CleanTrackTest(),
    WhiteNoiseTest(snr_db=0),
    PitchShiftTest(n_steps=2),
    DistortionTest(drive_db=20),
    ReverbTest(room_size=0.85, damping=0.6, wet_level=0.7, dry_level=0.3),
    ClippingDbTest(threshold_db=-20),
]

benchmark_runner.run_experiment(test_cases=test_cases_to_run)

results_df = benchmark_runner.get_results_df()
benchmark_runner.summarize_results()
```
