```python
import shutil
from sklearn.model_selection import train_test_split

from src.fma_utils import create_dataset_df

from src.TestSuite.test_config import TestConfig
from src.TestSuite.experiment_runner import ExperimentRunner
from src.TestSuite.test_cases import CleanTrackTest, WhiteNoiseTest, PitchShiftTest, ReverbTest, ClippingDbTest, DistortionTest

from src.fingerprinter import SVDFingerprinter, DeltaFingerprint
from src.fingerprint_database import HammingSearch, FFTConvolveSearch
```

```python
METADATA_DIR = "B:\\Documents\\Uni\\MATH3092 Mathematics Project\\MusicDatasets\\fma\\fma_metadata"
USER_SONGS_DIR = "B:\\Documents\\Uni\\MATH3092 Mathematics Project\\MusicDatasets\\my_songs" # random extra songs that aren't part of the FMA dataset
AUDIO_FILES_DIR = "B:\\Documents\\Uni\\MATH3092 Mathematics Project\\MusicDatasets\\fma\\fma_small\\fma_small"

tracks, my_songs = create_dataset_df(METADATA_DIR, AUDIO_FILES_DIR, USER_SONGS_DIR)
```

```python
f_s = 44100
L = 2048 * 2
H = 1024 * 2
window_type = "hann"
B = 128
tau = 70
gamma = 50000
kappa = 500
phi = 16
random_seed = 42
match_threshold = 0.35
```

```python
test_size = 0.5
num_tracks = 8000

reduced_track_db = tracks.iloc[0:num_tracks]
X_train, X_test = train_test_split(
    reduced_track_db,
    test_size=test_size,
    random_state=random_seed
)

fingerprinter = SVDFingerprinter(
    fingerprint_strategy=DeltaFingerprint(),
    f_s=f_s,
    L=L,
    H=H,
    B=B,
    tau=tau,
    window_type=window_type,
    gamma=gamma,
    kappa=kappa,
    phi=phi,
    random_state=random_seed
)

fingerprinter.train(X_train['filepath'].tolist())
```

```python
# Create the test config
config = TestConfig(
    f_s=f_s,
    snippet_duration_sec=5.0,
    match_threshold=match_threshold,
    n_db_tracks=100,
    n_query_tracks=50,
    random_seed=random_seed,
    save_failed_queries=True,
    save_passed_queries=True,
    temp_dir="temp_queries"
)

# Create the experiment runner
runner = ExperimentRunner(
    fingerprinter=fingerprinter,
    search_strategy=FFTConvolveSearch(),
    all_filepaths=X_test['filepath'].tolist(),
    config=config
)

runner.setup_database()
```

```python
test_cases_to_run = [
    # CleanTrackTest(),
    # WhiteNoiseTest(amplitude=0.05),
    # WhiteNoiseTest(amplitude=0.10),
    # WhiteNoise(amplitude=0.15),
    # PitchShiftTest(n_steps=0.5), # Shift up by half a semitone
    DistortionTest(drive_db=20),
    ReverbTest(room_size=0.85, damping=0.6, wet_level=0.7, dry_level=0.3),
    ClippingDbTest(threshold_db=-12),
]

runner.run_experiment(test_cases=test_cases_to_run)
results_df = runner.get_results_df()

print("\n--- DETAILED RESULTS ---")
print(results_df.to_string())

runner.summarize_results(results_df) # 049029
```
